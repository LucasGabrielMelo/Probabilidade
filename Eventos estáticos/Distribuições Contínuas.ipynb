# Variáveis aleatórias Continuas #

Uma variável aleatória contínua é aquela que assume valores em um intervalo contínuo $\mathcal{I} = [a,b)$, que pode ser limitado ou não. Dada uma variável aleatória contínua $X$, ao invés de calcularmos a probabilidade de ela assumir um valor $x$ específico, calculamos a probabilidade de ela assumir um valor pertencente ao intervalo $[x_0,x_1) \subset \mathcal{I}$, para isso utilizamos a **função densidade de probabilidades de X**, denotada por $f_X(x)$. Conhecida essa função, temos que

$$P[x_0 \le X < x_1] = \int_{x_0}^{x_1}f_X(x)dx\,.$$

Para essa variável, a média $E[X]$ e a variância $Var(X)$ são dadas por

$$E[X] = \mu = \int_{-\infty}^{\infty}x\cdot f_X(x)dx\,,$$
$$Var(X)=\sigma^2=E[(X-\mu)^2] = E[X^2] - {E[X]}^2\,.$$

### Distribuição uniforme ###

Uma distribuição contínua uniforme de probabilidades é uma distribuição em que uma variável aleatória $X$ toma valores em um intervalo finito $[a.b)$ de maneira uniforme, isto é, a probabilidade de $X$ tomar valores em um intervalo $[x_0,x_1) \subset [a,b)$ é proporcional ao comprimento desse intervalo. Nesse tipo de variável, tem-se

$$
f_X(x) =\begin{cases}
\dfrac{1}{b-a}, & a \leq x \leq b,\\
0, & \text{caso contrário}.
\end{cases}\,,
$$

$$E[X] = \frac{b+a}{2}\,$$
$$Var(X) = \frac{(b-a)^2}{12}\,.$$

Se uma variável aleatória $X$ segue essa distribuição de probabilidades, escreve-se $X \sim U[a,b)$.
