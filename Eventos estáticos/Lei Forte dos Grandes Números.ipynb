```python
import numpy as np
import matplotlib.pyplot as plt
import eventos
```

# Lei Forte dos Grandes Números #

Dado um conjunto de $N$ variáveis aleatórias independentes e identicamente distribuídas (i.i.d.) com média $\mu$, $\{X_1,X_2,...,X_N\}$, definindo-se a variável aleatória Média Amostral,

$$\overline{X}_N = \frac{1}{N}\sum\limits_{i=1}^{N}X_i\,,$$

a **Lei Forte dos Grandes Números** garante que, à medida que $N$ aumenta ($N \to \infty$), o valor da Média Amostral se aproxima quase certamente de $\mu$. Essa lei é representada de forma matemática pela equação

$$P\left[\,\lim_{N\to\infty}\overline{X}_N = \mu\right] = 1\,.$$

### Variáveis aleatórias de Bernoulli ###

```python
X_medio_array = []

p = 0.6

N = 100
for k in range(1,N+1,1):
    X_medio = 0
    for j in range(0,k):
        b = eventos.Bernoulli(p)
        X_medio += b/k
    X_medio_array.append(X_medio)
X_medio_array = np.array(X_medio_array)

plt.plot(np.arange(1,N+1,1), X_medio_array, color = 'blue')
plt.axhline(p, linestyle = '--', color = 'red', label = 'Média das variáveis de Bernoulli')
plt.grid(True)
plt.title(r'Valor médio de diversos experimentos de Bernoulli ($p=%.2f$)' %p)
plt.legend(loc = 'best')
plt.xlabel('Número de experimentos')
plt.ylabel('Média Amostral')
plt.show()
```

### Variáveis aleatórias binomiais ###

```python
X_medio_array = []

p = 0.4 # Probabilidade de sucesso
n = 10 # 10 repetições

N = 80
for k in range(1,N+1,1):
    X_medio = 0
    for j in range(0,k):
        b = eventos.Bin(n,p)
        X_medio += b/k
    X_medio_array.append(X_medio)
X_medio_array = np.array(X_medio_array)

plt.plot(np.arange(1,N+1,1), X_medio_array, color = 'blue')
plt.axhline(n*p, linestyle = '--', color = 'red', label = 'Média das variáveis binomiais')
plt.grid(True)
plt.title(r'Valor médio de diversos experimentos binomiais ($n=%d\,,\,p=%.2f$)' %(n,p))
plt.legend(loc = 'best')
plt.xlabel('Número de experimentos')
plt.ylabel('Média Amostral')
plt.show()
```

### Variáveis aleatórias de Poisson ###

```python
X_medio_array = []

lambda_ = 5 # Probabilidade de sucesso

N = 80
for k in range(1,N+1,1):
    X_medio = 0
    for j in range(0,k):
        p = eventos.Poisson(lambda_)
        X_medio += p/k
    X_medio_array.append(X_medio)
X_medio_array = np.array(X_medio_array)

plt.plot(np.arange(1,N+1,1), X_medio_array, color = 'blue')
plt.axhline(lambda_, linestyle = '--', color = 'red', label = 'Média das variáveis de Poisson')
plt.grid(True)
plt.title(r'Valor médio de diversos experimentos de Poisson ($\lambda=%d$)' %(lambda_))
plt.legend(loc = 'best')
plt.xlabel('Número de experimentos')
plt.ylabel('Média Amostral')
plt.show()
```

### Variáveis aleatórias com distribuição de probabilidades qualquer ###

```python
'''
Coloque uma distribuição de probabilidade qualquer aqui.

Distribuição utilizada: X assume valores no conjunto {-1,0,1}, com P[X=-1] = 1/4, P[X=0] = 1/4 e P[X=1] = 1/2.
'''

I = np.array([-1,0,1])
p_array = np.array([1/4,1/4,1/2])

def X():
    return np.random.choice(I, p=p_array)

EX = 0

for k in range(0, len(I), 1): # Calcula a média
    xi = I[k]
    pi = p_array[k]
    EX += xi*pi

## Realizando os experimentos

X_medio_array = []

N = 80
for k in range(1,N+1,1):
    X_medio = 0
    for j in range(0,k):
        x = X()
        X_medio += x/k
    X_medio_array.append(X_medio)
X_medio_array = np.array(X_medio_array)

plt.plot(np.arange(1,N+1,1), X_medio_array, color = 'blue')
plt.axhline(EX, linestyle = '--', color = 'red', label = 'Média das variáveis aleatórias')
plt.grid(True)
plt.title(r'Valor médio de diversos experimentos da variável aleatória $X$')
plt.legend(loc = 'best')
plt.xlabel('Número de experimentos')
plt.ylabel('Média Amostral')
plt.show()
```
